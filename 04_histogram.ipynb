# Histogram

Dalam sesi ini kita akan mempelajari cara membuat histogram dengan Matplotlib.

## Import modules

```python
%matplotlib inline
```

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print(matplotlib.__version__)
print(np.__version__)
```

## Sample Data

```python
coba = np.random.randn(10000)
coba
```

```python
plt.hist(coba, 
         bins=50, 
         facecolor='g', 
         alpha=0.50)

plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')
plt.title('Contoh Histogram')

plt.text(-3, 500, r'$\mu=0,\ \sigma=1$')
plt.grid()

plt.show()
```

```python
mu, sigma = 100, 15 # nilai mean dan nilai standard deviation

x = mu + sigma * np.random.randn(10000) # normal distribution
x

# np.random.randn(10000) membuat data random ber dist normal disekitar 0
# kita mau geser mean nya jadi 100
# dan menambahkan std jadi 15 (melebar)
```

```python
x.shape
```

## Histogram dengan pyplot Style

```python
plt.hist(x, 
         bins=50, 
         facecolor='g', 
         alpha=0.50)

plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')
plt.title('Contoh Histogram')

plt.text(45, 500, r'$\mu=100,\ \sigma=15$')
plt.grid()

plt.show()

# bins untuk jumlah kelas frekusensi
# alpha untuk transparansi diagram
```

## Histogram dengan OO Style

```python
fig, ax = plt.subplots()

ax.hist(x, 
        bins=50, 
        facecolor='b', 
        alpha=0.25)

ax.set(xlabel = 'Sumbu X',
       ylabel = 'Sumbu Y',
       title = 'Contoh Histogram')

ax.text(45, 500, r'$\mu=100,\ \sigma=15$')
ax.grid()

plt.show()
```
