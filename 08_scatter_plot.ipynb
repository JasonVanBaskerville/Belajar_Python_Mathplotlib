# Scatter Plot / Scatter Chart

Dalam sesi ini kita akan mempelajari cara membuat scatter plot dengan Matplotlib.

## Import modules

```python
%matplotlib inline
```

```python
import matplotlib
import matplotlib.pyplot as plt

print(matplotlib.__version__)
```

## Simple Scatter Plot

```python
x = [2,4,6,8,10,11,11.5,11.7]
y = [1,1.5,2,2.5,3,3.5,4,4.5]
```

```python
plt.scatter(x, y, label='Data 1', color='r')

plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')
plt.title('Contoh Scatter Plot')

plt.legend()

plt.show()
```

## Multiple Scatter Plot

```python
x1 = [2,4,6,8,10,11,11.5,11.7]
y1 = [1,1.5,2,2.5,3,3.5,4,4.5]

x2 = [8,8.5,9,9.5,10,10.5,11]
y2 = [3,3.5,3.7,4,4.5,5,5.2]

print(x1,y1,x2,y2)
```

```python
plt.scatter(x1, y1, label='Data 1', color='r')
plt.scatter(x2, y2, label='Data 2', color='b')

plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')
plt.title('Multiple Scatter Plot')

plt.legend()
plt.show()
```

## Pengaturan Marker

```python
x1 = [2,4,6,8,10,11,11.5,11.7]
y1 = [1,1.5,2,2.5,3,3.5,4,4.5]

x2 = [8,8.5,9,9.5,10,10.5,11]
y2 = [3,3.5,3.7,4,4.5,5,5.2]
```

```python
plt.scatter(x1, y1, 
            color='cyan',  
            linewidths=2,  # tebal luar 
            marker='s',   # square/kotak
            edgecolor='black',    # sisi luar
            s=400)    # besar kotak
  
plt.scatter(x2, y2, 
            color='yellow', 
            linewidths=1, 
            marker='^',  
            edgecolor='blue',  
            s=200) 

plt.show()
```

## Scatter Plot untuk data multi dimensi

```python
from sklearn.datasets import load_iris
iris = load_iris()
features = iris['data'].T
```

```python
iris.keys()
```

```python
iris['data']
```

```python
iris['feature_names']
```

```python
features = iris['data'].T
features
```

```python
iris['target']
```

```python
iris['target_names']
```

```python
plt.scatter(features[0], features[1], # x = sepallength, y = sepalwidth
            alpha=0.5,  # tranparansi
            s=100*features[3],  # fitur ke-3, semakin lebar lingkaran, semakin besar nilai petalwidth-nya
            c=iris['target'],   # c ini label target = setosa, versicolor, virginica
            cmap='viridis')     # kelompokkan warna bedasarkan c

plt.xlabel(iris['feature_names'][0])
plt.ylabel(iris['feature_names'][1])
plt.title('Iris Dataset')

plt.colorbar()   # tampilan color bar

plt.show()
```
