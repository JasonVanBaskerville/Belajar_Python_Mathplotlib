# Pie Plot / Pie Chart

Dalam sesi ini kita akan mempelajari cara membuat pie plot dengan Matplotlib.

## Import modules

```python
%matplotlib inline
```

```python
import matplotlib
import matplotlib.pyplot as plt

print(matplotlib.__version__)
```

## Simple Pie Plot

```python
kategori = ['A', 'B', 'C', 'D', 'E']
data = [215, 130, 245, 210, 225]
```

```python
plt.pie(data, 
        labels=kategori,
        autopct='%1.1f%%',  # 1.0 = 0 angka dibekang koma, 1.1 = 1 angka dibelakang koma, dst
        startangle=90)  # mulai dari sudur 90 derajat

plt.title('Contoh Pie Plot')

plt.show()
```

## Pengaturan Legend dan Color

```python
kategori = ['A', 'B', 'C', 'D', 'E']
data = [215, 130, 245, 210, 225]
warna = ['pink', 'cyan', '#e03364', 'yellowgreen', 'skyblue']
```

```python
plt.pie(data, 
        labels=kategori,
        colors=warna,
        autopct='%1.1f%%',
        startangle=90)

plt.legend()
plt.title('Contoh Pie Plot')

plt.show()
```

## Exploded Pie Plot

```python
kategori = ['Kota_A', 'Kota_B', 'Kota_C', 'Kota_D', 'Kota_E']
data = [215, 130, 245, 210, 225]
warna = ['pink', 'cyan', '#e03364', 'yellowgreen', 'skyblue']
explode_var = [0., 0.2, 0., 0., 0.]
```

```python
plt.pie(data, 
        labels=kategori,
        colors=warna,
        explode=explode_var,
        autopct='%1.1f%%',
        startangle=90)

plt.title('Contoh Exploded Pie Plot')

plt.show()
print('Hasil survey kota B menunjukkan bahwa 12.7% warga masuk kategori desil 10')
```
