# Box Plot / Box Chart

Dalam sesi ini kita akan mempelajari cara membuat box plot dengan Matplotlib.

## Import modules

```python
%matplotlib inline
```

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np

print(matplotlib.__version__)
print(np.__version__)
```

## Sample Dataset

```python
np.random.seed(2)  # hasil random sama setiap pengulangan untuk seed = 2, boleh seed = 1 dll
data = np.random.normal(loc=100, scale=10, size=200)  # (mean, std, banyak data) 
data

# bedanya dengan random.randn adalah mean = 0 dan std = 1 (normal standar)
```

## Simple Box Plot

```python
## DEPRECATED
# plt.boxplot(data, labels=['Data'])

# plt.title('Simple Box Plot')
# plt.xlabel('Sumbu X')
# plt.ylabel('Sumbu Y')

# plt.grid()
# plt.show()
```

```python
plt.boxplot(data, tick_labels=['Data'])

plt.title('Simple Box Plot')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.grid()
plt.show()

# garis kuning = median/Q2
# sisi kotak = Q1 dan Q3
# garis atas = nilai max yg masih dalam jangkauan Q_3 + (1.5 x IQR)
# garis bawah = nilai min yg masih dalam jangkauan Q_1 - (1.5 x IQR)
# IQR = Q3 - Q1
```

## Notched Box Plot

```python
plt.boxplot(data, tick_labels=['Data'], notch=True)

plt.title('Notched Box Plot')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.show()
```

## Pengaturan marker untuk outlier

```python
plt.boxplot(data, 
            tick_labels=['Data'],
            showfliers=True,  # if false, outlier tidak muncul 
            flierprops={'markerfacecolor':'r', 'marker':'d'})   # r = red, d = diamond

plt.title('Pengaturan Marker')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.show()
```

## Horizontal Box Plot

```python
plt.boxplot(data, tick_labels=['Data'], vert=False)  # if true, box plot vertikal

plt.title('Horisontal Box Plot')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.show()
```

## Multiple Box Plot

```python
np.random.seed(2)
data1 = np.random.normal(100, 10, 200)
data2 = np.random.normal(80, 30, 200)
data3 = np.random.normal(90, 20, 200)
data4 = np.random.normal(70, 25, 200)

data = [data1, data2, data3, data4] 
labels = ['Data 1', 'Data 2', 'Data 3', 'Data 4']
```

```python
## DEPRECATED
# plt.boxplot(data, labels=labels)

# plt.title('Multiple Box Plot')
# plt.xlabel('Sumbu X')
# plt.ylabel('Sumbu Y')

# plt.show()
```

```python
plt.boxplot(data, tick_labels=labels)  

plt.title('Multiple Box Plot')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.show()
```
