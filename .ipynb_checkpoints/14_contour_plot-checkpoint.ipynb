# Contour Plot

Dalam sesi ini kita akan mempelajari cara membuat contour plot dengan Matplotlib.

## Import modules

```python
%matplotlib inline
```

```python
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt

print(matplotlib.__version__)
print(np.__version__)
```

## Sample dataset

```python
# contoh kontur mulus (sulit)
# n = np.linspace(-3.0, 3.0, 100)

# x, y = np.meshgrid(n, n)
# z = np.sqrt(x**2 + y**2)

# print(x.shape)
# print(y.shape)
# print(z.shape)

# contoh kontur mudah
n = [-1, 0, 1]
x, y = np.meshgrid(n, n)
# x =
# [[1 2 3]
#  [1 2 3]
#  [1 2 3]]

# y =
# [[1 1 1]
#  [2 2 2]
#  [3 3 3]]

z = np.sqrt(x**2 + y**2)
# z =
# [[1²+1², 2²+1², 3²+1²],
#  [1²+2², 2²+2², 3²+2²],
#  [1²+3², 2²+3², 3²+3²]]
```

```python
df_x = pd.DataFrame(x)
df_x
```

```python
df_y = pd.DataFrame(y)
df_y
```

```python
df_z = pd.DataFrame(z)
df_z
```

## Contour Plot

```python
plt.contour(x, y, z)
plt.colorbar()

plt.title('Contour Plot')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.show()
```

## Filled Contour Plot

```python
plt.contourf(x, y, z)
plt.colorbar()

plt.title('Filled Contour Plot')
plt.xlabel('Sumbu X')
plt.ylabel('Sumbu Y')

plt.show()
```
